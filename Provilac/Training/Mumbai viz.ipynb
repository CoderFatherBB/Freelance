```python
import pandas as pd
df = pd.read_excel('Mumbai Server data 01 Jan 2020- 30th Aug.xlsx')
```

```python
import matplotlib.pyplot as plt

# Convert 'DeliveryDate' to datetime format
df['delivery_date'] = pd.to_datetime(df['delivery_date'],format="%d-%m-%Y")

# Get unique product names
products = df['product'].unique()

# Create a plot for each product
for product in products:
    product_df = df[df['product'] == product]  # Filter data for the current product
    product_df = product_df.sort_values('delivery_date')  # Sort data by date
    
    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(product_df['delivery_date'], product_df['total_quantity'], marker='o', linestyle='-')
    plt.title(f'Trend of Total Quantity for {product} (2020-2024)')
    plt.xlabel('Date')
    plt.ylabel('Total Quantity')
    plt.xticks(rotation=45)  # Rotating x-axis labels for better readability
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
