```python
c = c[c != 'Chocolate Milk 300ml']
c
```

```python
All = pd.DataFrame(columns=['Name', 'Train Results', 'Total Results'])

def save_file_with_dirs(filepath,obj):
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    joblib.dump(obj, filepath)

count = 0
for _ in c:
    print(_)
    
    a=pd.read_csv("Hyderabad item type wise product data- 01 Jan 2020- 10 April 2024.csv")
    a=a[a["Product Name"]==_]
    
    a['DeliveryDate'] = pd.to_datetime(a['DeliveryDate'],format="%d-%m-%Y")
    # Sorting the DataFrame by date
    a = a.sort_values('DeliveryDate')

    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(a['DeliveryDate'], a['Total Normal Quantity'], marker='o', linestyle='-')
    plt.title('Trend of Total Normal Quantity (2020-2024)')
    plt.xlabel('Date')
    plt.ylabel('Total Normal Quantity')
    plt.xticks(rotation=45)  # Rotating x-axis labels for better readability
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    a = a[a['DeliveryDate'].dt.year == 2022]

    plt.figure(figsize=(10, 6))
    plt.plot(a['DeliveryDate'], a['Total Normal Quantity'], marker='o', linestyle='-')
    plt.title('Trend of Total Normal Quantity (2020-2024)')
    plt.xlabel('Date')
    plt.ylabel('Total Normal Quantity')
    plt.xticks(rotation=45)  # Rotating x-axis labels for better readability
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    a['DeliveryDate'] = pd.to_datetime(a['DeliveryDate'],format="%d-%m-%Y")
    a['year'] = a['DeliveryDate'].dt.year
    a['month'] = a['DeliveryDate'].dt.month
    a['day'] = a['DeliveryDate'].dt.day


    X=a.drop(columns=["DeliveryDate","City","Product Name","Total Normal Quantity","Total Coupon Quantity","Total Free Quantity"])
    y=a[["Total Normal Quantity"]]

    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()
    X_scaled = scaler_X.fit_transform(X)
    y_scaled = scaler_y.fit_transform(y)

    X_train,X_test,y_train,y_test = train_test_split(X_scaled,y_scaled,test_size=0.2,random_state=16)


    model = RandomForestRegressor()
    model.fit(X_train,y_train)

    z1 = model.score(X_test, y_test)
    z2 = model.score(X_scaled,y_scaled)
    print (z1, z2)

    All.loc[count] = [_, z1*100, z2*100]
    count = count + 1

    # Save the model and scalers
    save_file_with_dirs(f"Saved/Hyderabad/{_}.joblib", model)
    save_file_with_dirs(f"Scalers/Hyderabad/{_}/{_}x.save", scaler_X)
    save_file_with_dirs(f"Scalers/Hyderabad/{_}/{_}y.save", scaler_y)

```
