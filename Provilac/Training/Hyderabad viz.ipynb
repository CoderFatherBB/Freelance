```python
import pandas as pd

df = pd.read_csv('Hyderabad item type wise product data- 01 Jan 2020- 10 April 2024.csv')
```

```python
df['Product Name'].unique()
```

```python
df['Product Name'].value_counts()
```

```python
import matplotlib.pyplot as plt

# Convert 'DeliveryDate' to datetime format
df['DeliveryDate'] = pd.to_datetime(df['DeliveryDate'],format="%d-%m-%Y")

# Get unique product names
products = df['Product Name'].unique()

# Create a plot for each product
for product in products:
    product_df = df[df['Product Name'] == product]  # Filter data for the current product
    product_df = product_df.sort_values('DeliveryDate')  # Sort data by date
    
    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(product_df['DeliveryDate'], product_df['Total Normal Quantity'], marker='o', linestyle='-')
    plt.title(f'Trend of Total Normal Quantity for {product} (2020-2024)')
    plt.xlabel('Date')
    plt.ylabel('Total Normal Quantity')
    plt.xticks(rotation=45)  # Rotating x-axis labels for better readability
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```
