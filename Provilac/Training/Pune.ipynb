```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from datetime import datetime
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
import os
import joblib
```

```python
a=pd.read_excel('Pune Server Delivery data 01 Jan 2020- 30 Aug 2024.xlsx')
c = a["product"].unique()
```

```python
All = pd.DataFrame(columns=['Name', 'Train Results', 'Total Results'])

def save_file_with_dirs(filepath,obj):
    os.makedirs(os.path.dirname(filepath), exist_ok=True)
    joblib.dump(obj, filepath)

count = 0
for _ in c:
    print(_)
    
    a=pd.read_excel("Pune Server Delivery data 01 Jan 2020- 30 Aug 2024.xlsx")
    a=a[a["product"]==_]
    
    a['delivery_date'] = pd.to_datetime(a['delivery_date'],format="%d-%m-%Y")
    # Sorting the DataFrame by date
    a = a.sort_values('delivery_date')

    # Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(a['delivery_date'], a['total_quantity'], marker='o', linestyle='-')
    plt.title('Trend of Total Normal Quantity (2020-2024)')
    plt.xlabel('Date')
    plt.ylabel('Total Normal Quantity')
    plt.xticks(rotation=45)  # Rotating x-axis labels for better readability
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    a['delivery_date'] = pd.to_datetime(a['delivery_date'],format="%d-%m-%Y")
    a['year'] = a['delivery_date'].dt.year
    a['month'] = a['delivery_date'].dt.month
    a['day'] = a['delivery_date'].dt.day


    X=a.drop(columns=['City', 'delivery_date', 'product', 'total_quantity'])
    y=a[["total_quantity"]]

    scaler_X = MinMaxScaler()
    scaler_y = MinMaxScaler()
    X_scaled = scaler_X.fit_transform(X)
    y_scaled = scaler_y.fit_transform(y)

    X_train,X_test,y_train,y_test = train_test_split(X_scaled,y_scaled,test_size=0.2,random_state=16)


    model = RandomForestRegressor()
    model.fit(X_train,y_train)

    z1 = model.score(X_test, y_test)
    z2 = model.score(X_scaled,y_scaled)
    print (z1, z2)

    All.loc[count] = [_, z1*100, z2*100]
    count = count + 1

    # Save the model and scalers
    save_file_with_dirs(f"Saved/Pune/{_}.joblib", model)
    save_file_with_dirs(f"Scalers/Pune/{_}/{_}x.save", scaler_X)
    save_file_with_dirs(f"Scalers/Pune/{_}/{_}y.save", scaler_y)

```
