```python
import os
import numpy as np
from PIL import Image, ImageEnhance
from keras.preprocessing.image import ImageDataGenerator

input_directory = r'D:\Object_det\test'
output_directory = r'D:\Object_det\test'

SIZE = 224

# Initialize the ImageDataGenerator for other augmentations
# datagen = ImageDataGenerator(
#     rotation_range=15,
#     vertical_flip=True,
#     channel_shift_range=20,
#     brightness_range=(0.5, 1.5)
# )

# Create a function for color augmentation using Pillow

for root, dirs, files in os.walk(input_directory):
    for dir_name in dirs:
        input_subfolder = os.path.join(root, dir_name)
        output_subfolder = os.path.join(output_directory, os.path.relpath(input_subfolder, input_directory))

        if not os.path.exists(output_subfolder):
            os.makedirs(output_subfolder)

        my_images = os.listdir(input_subfolder)

        for image_name in my_images:
            if image_name.split('.')[-1] == 'jpg' or image_name.split('.')[-1] == 'JPG':
                image = Image.open(os.path.join(input_subfolder, image_name))
                
                # Get original image size
                original_size = image.size
                
                # Rotate the image without expanding
                rotated_image1 = image.rotate(15, expand=False)
                rotated_image3 = image.rotate(-15, expand=False)
                
                # Apply vertical flip
                flipped_image = rotated_image1.transpose(Image.FLIP_TOP_BOTTOM)
                
                # Rotate again without expanding
                rotated_image2 = flipped_image.rotate(15, expand=False)
                rotated_image4 = flipped_image.rotate(-15, expand=False)
                
                # Save the augmented images
                rotated_image1.save(os.path.join(output_subfolder, f"rotated1_{image_name}"))
                rotated_image3.save(os.path.join(output_subfolder, f"rotated3_{image_name}"))  # Saving -15 degrees rotation
                flipped_image.save(os.path.join(output_subfolder, f"flipped_{image_name}"))
                rotated_image2.save(os.path.join(output_subfolder, f"rotated2_{image_name}"))
                rotated_image4.save(os.path.join(output_subfolder, f"rotated4_{image_name}"))  # Saving -15 degrees rotation

print('Complete')
```
